# Task 2

```python
from pyspark.sql.functions import max, min, lit

# code for getting MAX and MIN price
row_count = df.count()
stats = df.select(min("price"), max("price")).collect()[0]

min_price = stats[0]
max_price = stats[1]

# Formatting the content as text
content = f"min_price,max_price,row_count\n{min_price},{max_price},{row_count}"

# Writing to text file
dbutils.fs.put(
    "/Workspace/test-assignment/out/out_2_2.txt",
    content,
    overwrite=True
)

print("Text file created successfully")
```

# Task 3

```python
from pyspark.sql.functions import avg
filtered_df = df.filter((df['price'] > 5000) & (df['review_scores_value'] == 10))
avg_bathrooms = filtered_df.select(avg("bathrooms")).collect()[0][0]
avg_bedrooms = filtered_df.select(avg("bedrooms")).collect()[0][0]

content_avg = f"avg bathrooms,avg bedrooms\n{avg_bathrooms},{avg_bedrooms}"

# Writing to text file
dbutils.fs.put(
    "/Workspace/test-assignment/out/out_2_3.txt",
    content_avg,
    overwrite=True
)

print("Text file created successfully")
```

# task 4

```python
from pyspark.sql.functions import min, max, sum

#storing the minimum price
min_price = df.select(min("price")).collect()[0][0]

#storing the highest rating
highest_rating = df.select(max("review_scores_rating")).collect()[0][0]

# filtering the dataframe for properties with lowest price and highest rating
prop_lowprice_highrating = df.filter((df["review_scores_rating"] == highest_rating) & (df["price"] == min_price))

# storing the result in variable by summing up the accomedates column
lowest_price_highest_rating_accommodates = prop_lowprice_highrating.select(sum("accommodates")).collect()[0][0]

content_people = f"{lowest_price_highest_rating_accommodates}"

# Writing to text file
dbutils.fs.put(
    "/Workspace/test-assignment/out/out_2_4.txt",
    content_people,
    overwrite=True
)
```
